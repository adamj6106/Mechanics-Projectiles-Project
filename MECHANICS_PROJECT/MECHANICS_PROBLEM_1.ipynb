```python
from scipy import optimize

#Brents method is what is used here, Brent’s method combines root bracketing, 
#interval bisection, and inverse quadratic interpolation. 

# Import packages
import numpy as np
from scipy.optimize import brentq

# Starting angle
theta = 45

# Convert to radians
theta = np.radians(theta)


# Our equation
def f(R):
    return (R - np.sin(2*(theta)))

# Interval setting for the root finding
R = optimize.brentq(f, 0, 1)

print ( "Range" , R)
```

```python
#Same thing but with Linear air resistance insted

from scipy import optimize

import numpy as np
from scipy.optimize import brentq


theta = 0.75


def f(R):
    return ((np.sin(theta) + 1) / np.cos(theta)) * R + \
           np.log(1 - R / np.cos(theta))

R = optimize.brentq(f, 0.1, 0.9 * np.cos(theta))

print("Range:", R)
```

```python
import matplotlib.pyplot as plt
from scipy import optimize


import numpy as np
from scipy.optimize import brentq



theta = 0.62



def f(R):
    return ((np.sin(theta) + 1) / np.cos(theta)) * R + \
           np.log(1 - R / np.cos(theta))

R = optimize.brentq(f, 0.3, 0.7 * np.cos(theta))

print("Range", R)

#Plotting

# Horizontal positions from launch to landing
x = np.linspace(0, R, 100)

# Height with linear air resistance
y = (1 + np.sin(theta)) * x + \
    np.cos(theta) * np.log(1 - x / np.cos(theta))

# Plot the trajectory
plt.plot(x, y, label="Projectile Trajectory")

plt.xlabel("Horizontal Distance")
plt.ylabel("Height")
plt.title("Projectile Trajectory with Linear Air Resistance")

plt.grid(True)
plt.legend()
plt.show()
```

```python
#New plot comparing the air resistance vs vacuum trajectory

import matplotlib.pyplot as plt
from scipy import optimize


import numpy as np
from scipy.optimize import brentq



theta = 0.62



def f(R):
    return ((np.sin(theta) + 1) / np.cos(theta)) * R + \
           np.log(1 - R / np.cos(theta))


R = optimize.brentq(f, 0.3, 0.7 * np.cos(theta))

print("Range", R)

# Trajectory without air resistance
R_vacuum = np.sin(2 * theta)

x_vacuum = np.linspace(0, R_vacuum, 1000)

y_vacuum = x_vacuum * np.tan(theta) - \
    x_vacuum**2 / (2 * np.cos(theta)**2)


#Trajectory with Linear air resistance 
x = np.linspace(0, R, 100)


y = (1 + np.sin(theta)) * x + \
    np.cos(theta) * np.log(1 - x / np.cos(theta))

# Plot the trajectory
plt.plot(x, y, label="With Air Resistance", color="blue")
plt.plot(x_vacuum, y_vacuum,
         label="Without Air Resistance", color="red")

plt.xlabel("Range")
plt.ylabel("Height")
plt.title("Linear Air Resistance Vs. Withouth Air Resistance")

plt.grid(True)
plt.legend()
plt.show()
```
